```python
def modulo_caja_crud():
    print("=== PagaYa S.A.S. - Modulo de Control de Caja (CRUD) ===")

    cajero = input("Ingrese el codigo o nombre del cajero: ")
    while True:
        try:
            tope = float(input("Ingrese el tope maximo de recaudo (> 0): "))
            if tope > 0:
                break
            print("El tope debe ser mayor a cero.")
        except ValueError:
            print("Por favor, ingrese un numero valido.")

    transacciones = []
    recaudo_total = 0
    motivo_cierre = "Fin de cola"

    print(f"\nTurno iniciado para {cajero} con tope de ${tope:,.2f}")
    print("-" * 50)

    while recaudo_total < tope:
        print("\n--- MENU DE OPERACIONES (CRUD) ---")
        print("1. Registrar nueva transaccion (Crear)")
        print("2. Ver transacciones (Leer)")
        print("3. Actualizar una transaccion (Actualizar)")
        print("4. Eliminar una transaccion (Eliminar)")
        print("5. Finalizar turno manualmente (FIN)")

        opcion = input("Seleccione una opcion: ").strip().lower()

        if opcion == '1' or opcion == 'c':
            entrada = input("Ingrese el monto del cliente (o digite 'FIN' para terminar): ").strip()
            if entrada.upper() == 'FIN' or entrada == '0':
                break

            try:
                monto = float(entrada)
                if monto <= 0:
                    print("Monto rechazado: Debe ser mayor a cero.")
                    continue

                transacciones.append(monto)
                recaudo_total += monto
                print(f"Transaccion registrada. Acumulado actual: ${recaudo_total:,.2f}")

                if recaudo_total >= tope:
                    print("\nCAJA SUSPENDIDA: se alcanzo el tope de recaudo. Dirijase a tesoreria.")
                    motivo_cierre = "Tope alcanzado"
                    break

            except ValueError:
                print("Valor no numerico rechazado. Intente de nuevo.")

        elif opcion == '2' or opcion == 'r':
            if not transacciones:
                print("No hay transacciones registradas todavia.")
            else:
                print("\n--- LISTA DE TRANSACCIONES (LECTURA) ---")
                for i, t in enumerate(transacciones, 1):
                    print(f"Cliente {i}: ${t:,.2f}")
                print(f"Total acumulado: ${recaudo_total:,.2f}")

        elif opcion == '3' or opcion == 'u':
            if not transacciones:
                print("No hay transacciones para actualizar.")
            else:
                for i, t in enumerate(transacciones, 1):
                    print(f"{i}. ${t:,.2f}")
                try:
                    idx = int(input("Ingrese el numero de la transaccion a modificar: ")) - 1
                    if 0 <= idx < len(transacciones):
                        nuevo_monto = float(input("Ingrese el nuevo monto: "))
                        if nuevo_monto > 0:
                            recaudo_total = recaudo_total - transacciones[idx] + nuevo_monto
                            transacciones[idx] = nuevo_monto
                            print("Transaccion actualizada con exito.")

                            if recaudo_total >= tope:
                                print("\nCAJA SUSPENDIDA por superar el tope tras actualizacion.")
                                motivo_cierre = "Tope alcanzado"
                                break
                        else:
                            print("El monto debe ser mayor a cero.")
                    else:
                        print("Numero de transaccion invalido.")
                except ValueError:
                    print("Entrada invalida.")

        elif opcion == '4' or opcion == 'd':
            if not transacciones:
                print("No hay transacciones para eliminar.")
            else:
                for i, t in enumerate(transacciones, 1):
                    print(f"{i}. ${t:,.2f}")
                try:
                    idx = int(input("Ingrese el numero de la transaccion a eliminar: ")) - 1
                    if 0 <= idx < len(transacciones):
                        monto_eliminado = transacciones.pop(idx)
                        recaudo_total -= monto_eliminado
                        print(f"Transaccion de ${monto_eliminado:,.2f} eliminada con exito.")
                    else:
                        print("Numero de transaccion invalido.")
                except ValueError:
                    print("Entrada invalida.")

        elif opcion == '5' or opcion.upper() == 'FIN':
            motivo_cierre = "Fin de cola"
            break
        else:
            print("Opcion no valida. Intente de nuevo.")

    num_transacciones = len(transacciones)
    promedio = (recaudo_total / num_transacciones) if num_transacciones > 0 else 0.0

    print("\n" + "=" * 6 + " REPORTE DE TURNO " + "=" * 6)
    print(f"Cajero:               {cajero}")
    print(f"Tope asignado:        ${tope:,.2f}")
    print(f"Transacciones:        {num_transacciones}")
    print(f"Recaudo total:        ${recaudo_total:,.2f}")
    print(f"Promedio/transaccion: ${promedio:,.2f}")
    print(f"Motivo de cierre:     {motivo_cierre}")
    print("=" * 30)

modulo_caja_crud()
```
